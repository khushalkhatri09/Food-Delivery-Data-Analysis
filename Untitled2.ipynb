```python
# generate_dataset.py
import os
import numpy as np
import pandas as pd

# Reproducibility
np.random.seed(42)
N = 8000

# 1. IDs & Customers
order_ids = [f"ORD{str(i).zfill(6)}" for i in range(1, N + 1)]
customer_pool = [f"CUST{str(i).zfill(4)}" for i in range(1, 2201)]
customer_ids = np.random.choice(customer_pool, size=N, replace=True)

# 2. Dates & Times (Calendar Year 2025)
start_date = pd.Timestamp("2025-01-01")
end_date = pd.Timestamp("2025-12-31")
random_days = np.random.randint(0, (end_date - start_date).days + 1, size=N)
order_dates = [start_date + pd.Timedelta(days=int(d)) for d in random_days]

# Order hours: bimodal distribution peaking at Lunch (12-14) and Dinner (19-22)
hour_probs = np.array([
    0.01, 0.005, 0.005, 0.002, 0.003, 0.005, 0.01, 0.02, # 00:00 - 07:00
    0.03, 0.04,  0.05,  0.08,  0.11,  0.10,  0.05, 0.03, # 08:00 - 15:00
    0.03, 0.04,  0.07,  0.11,  0.11,  0.06,  0.02, 0.01  # 16:00 - 23:00
])
hour_probs /= hour_probs.sum()
order_hours = np.random.choice(np.arange(24), size=N, p=hour_probs)
order_minutes = np.random.randint(0, 60, size=N)
order_times = [f"{h:02d}:{m:02d}" for h, m in zip(order_hours, order_minutes)]

# 3. Categoricals
cities = ["Bengaluru", "Mumbai", "Delhi NCR", "Hyderabad", "Pune"]
city_probs = [0.28, 0.25, 0.22, 0.15, 0.10]
city_choices = np.random.choice(cities, size=N, p=city_probs)

food_categories = ["Biryani", "Pizza", "North Indian", "Burgers", "Chinese", "Desserts", "South Indian"]
cat_probs = [0.24, 0.20, 0.18, 0.14, 0.11, 0.07, 0.06]
food_choices = np.random.choice(food_categories, size=N, p=cat_probs)

restaurants = [
    "Meghana Foods", "Truffles", "Empire Restaurant", "Behrouz Biryani",
    "Domino's Pizza", "Burger King", "Haldiram's", "Wow! Momo",
    "The Belgian Waffle Co.", "Saravana Bhavan", "Mainland China", "KFC"
]
restaurant_choices = np.random.choice(restaurants, size=N)

payment_methods = ["UPI", "Credit Card", "Debit Card", "Cash on Delivery", "Net Banking"]
payment_probs = [0.55, 0.22, 0.10, 0.08, 0.05]
payment_choices = np.random.choice(payment_methods, size=N, p=payment_probs)

order_statuses = ["Delivered", "Cancelled", "Refunded"]
status_probs = [0.93, 0.05, 0.02]
status_choices = np.random.choice(order_statuses, size=N, p=status_probs)

# 4. Numerics & Metrics
items = np.random.choice([1, 2, 3, 4, 5, 6], size=N, p=[0.25, 0.35, 0.20, 0.12, 0.05, 0.03])
base_amount = items * np.random.uniform(140, 260, size=N) + np.random.exponential(scale=60, size=N)
order_amounts = np.round(base_amount, 2)

# Distance (km)
distances = np.round(np.random.gamma(shape=2.5, scale=1.8, size=N) + 0.5, 2)
distances = np.clip(distances, 0.8, 18.0)

# Dynamic Pricing Multiplier: applied during peak hours (12-14, 19-21) and high distances
is_peak = np.isin(order_hours, [12, 13, 19, 20, 21])
dynamic_price = np.where(is_peak, np.random.choice([1.15, 1.25, 1.35], size=N, p=[0.5, 0.35, 0.15]), 1.00)

# Delivery Fee: Rs 25 base + distance component * dynamic multiplier
delivery_fees = np.round((25.0 + distances * 4.5) * dynamic_price, 2)

# Discounts: Promotions offer 0%, 10%, 15%, or 20% on the order amount
discount_pct = np.random.choice([0.0, 0.10, 0.15, 0.20], size=N, p=[0.35, 0.30, 0.20, 0.15])
discounts = np.round(order_amounts * discount_pct, 2)

# Final Amount
final_amounts = np.round(order_amounts + delivery_fees - discounts, 2)

# Delivery Time (Minutes): base + distance * rate + peak traffic delay + noise
traffic_delay = np.where(is_peak, np.random.uniform(8, 16, size=N), np.random.uniform(2, 6, size=N))
delivery_times = np.round(15 + (distances * 2.8) + traffic_delay + np.random.normal(0, 3, size=N))
delivery_times = np.clip(delivery_times, 14, 85).astype(int)

# Customer Ratings (1-5): Slower deliveries degrade rating probabilities
ratings = []
for dt in delivery_times:
    if dt < 30:
        p = [0.01, 0.02, 0.07, 0.30, 0.60]
    elif dt <= 45:
        p = [0.03, 0.07, 0.20, 0.45, 0.25]
    else:
        p = [0.25, 0.35, 0.25, 0.10, 0.05]
    ratings.append(np.random.choice([1, 2, 3, 4, 5], p=p))

df = pd.DataFrame({
    "Order_ID": order_ids,
    "Customer_ID": customer_ids,
    "Order_Date": [d.strftime("%Y-%m-%d") for d in order_dates],
    "Order_Time": order_times,
    "City": city_choices,
    "Restaurant": restaurant_choices,
    "Food_Category": food_choices,
    "Number_of_Items": items,
    "Order_Amount": order_amounts,
    "Delivery_Fee": delivery_fees,
    "Discount": discounts,
    "Final_Amount": final_amounts,
    "Delivery_Time_Minutes": delivery_times,
    "Distance_KM": distances,
    "Customer_Rating": ratings,
    "Payment_Method": payment_choices,
    "Order_Status": status_choices,
    "Dynamic_Price": dynamic_price
})

# Classify customer type based on prior appearances
cust_order_counts = df.groupby("Customer_ID").cumcount()
df["Customer_Type"] = np.where(cust_order_counts > 0, "Returning", "New")

# Introduce a small amount of realistic dirty data to demonstrate cleaning in Step 4
# 1. 25 missing ratings (unrated orders)
nan_idx = np.random.choice(df.index, size=25, replace=False)
df.loc[nan_idx, "Customer_Rating"] = np.nan

# 2. 10 duplicate rows
duplicates = df.sample(10, random_state=42)
df = pd.concat([df, duplicates], ignore_index=True)

os.makedirs("data", exist_ok=True)
df.to_csv("data/food_delivery.csv", index=False)
print(f"Dataset generated successfully at data/food_delivery.csv. Total rows: {len(df)}")
```
